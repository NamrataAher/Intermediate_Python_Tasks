```python

import requests
from bs4 import BeautifulSoup
import csv
from urllib.parse import urljoin

# ShadowFox website
url = "https://www.shadowfox.in/"

try:
    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = requests.get(
        url,
        headers=headers,
        timeout=20
    )
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # 1. Extract website title
    title = (
        soup.title.get_text(strip=True)
        if soup.title else "No title found"
    )

    print("WEBSITE TITLE:", title)

    # 2. Extract headings
    headings = []

    for tag in soup.find_all(["h1", "h2", "h3"]):
        text = tag.get_text(" ", strip=True)

        if text:
            headings.append(text)

    print("\nWEBSITE HEADINGS:")
    for heading in headings:
        print("-", heading)

    # 3. Extract paragraphs
    paragraphs = []

    for tag in soup.find_all("p"):
        text = tag.get_text(" ", strip=True)

        if text:
            paragraphs.append(text)

    print("\nWEBSITE PARAGRAPHS:")
    for paragraph in paragraphs:
        print("-", paragraph)

    # 4. Extract links
    links = []

    for tag in soup.find_all("a", href=True):
        link_text = tag.get_text(" ", strip=True)
        link_url = urljoin(url, tag["href"])

        if link_url.startswith(("http://", "https://")):
            links.append({
                "text": link_text,
                "url": link_url
            })

    print("\nWEBSITE LINKS:")
    for link in links:
        print(link["text"], "-", link["url"])

    # 5. Save all data to CSV
    with open(
        "shadowfox_scraped_data.csv",
        "w",
        newline="",
        encoding="utf-8"
    ) as file:

        writer = csv.writer(file)
        writer.writerow(["Data Type", "Text", "URL"])

        writer.writerow(["Title", title, ""])

        for heading in headings:
            writer.writerow(["Heading", heading, ""])

        for paragraph in paragraphs:
            writer.writerow(["Paragraph", paragraph, ""])

        for link in links:
            writer.writerow([
                "Link",
                link["text"],
                link["url"]
            ])

    print("\nData saved to shadowfox_scraped_data.csv")
    print("Total headings:", len(headings))
    print("Total paragraphs:", len(paragraphs))
    print("Total links:", len(links))

except requests.RequestException as error:
    print("Website request failed:", error)

except Exception as error:
    print("An error occurred:", error)
```
