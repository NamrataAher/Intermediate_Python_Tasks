```python

import random

# Hangman Game
words = {
    "PYTHON": "A popular programming language",
    "COMPUTER": "An electronic machine",
    "PROGRAM": "A set of instructions",
    "KEYBOARD": "Used to type on a computer",
    "INTERNET": "A worldwide network"
}

word = random.choice(list(words.keys()))
guessed_letters = []
wrong_guesses = 0
max_attempts = 6

hangman_stages = [
    """
     +---+
     |   |
         |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
     |   |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|   |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    /    |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    / \\  |
    =========
    """
]

print("================================")
print("       HANGMAN WORD GAME")
print("================================")
print("Hint:", words[word])
print("Guess the word one letter at a time!")

while wrong_guesses < max_attempts:
    display_word = " ".join(
        letter if letter in guessed_letters else "_"
        for letter in word
    )

    print("\nWord:", display_word)
    print("Wrong guesses:", wrong_guesses, "/", max_attempts)
    print(hangman_stages[wrong_guesses])

    if all(letter in guessed_letters for letter in word):
        print("Congratulations! You guessed the word:", word)
        break

    guess = input("Enter one letter: ").strip().upper()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter only one alphabet letter.")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter. Try another!")
        continue

    guessed_letters.append(guess)

    if guess in word:
        print("Correct guess!")
    else:
        wrong_guesses += 1
        print("Wrong guess!")

else:
    print(hangman_stages[max_attempts])
    print("Game over! The correct word was:", word)

print("\nThanks for playing Hangman!")
```
